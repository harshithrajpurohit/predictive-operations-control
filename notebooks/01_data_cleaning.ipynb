```python
df = df.dropna(subset=['source_name', 'destination_name'])
df.shape
```

```python
trip_df = df.groupby('trip_uuid').agg({
    'route_type': 'first',
    'trip_creation_time': 'first',
    'source_center': 'first',
    'source_name': 'first',
    'destination_center': 'last',
    'destination_name': 'last',
    'od_start_time': 'first',
    'od_end_time': 'last',
    'start_scan_to_end_scan': 'sum',
    'actual_distance_to_destination': 'sum',
    'actual_time': 'sum',
    'osrm_time': 'sum',
    'osrm_distance': 'sum',
    'segment_actual_time': 'sum',
    'segment_osrm_time': 'sum',
    'segment_osrm_distance': 'sum'
}).reset_index()

trip_df.shape
```

```python
import re

# Extract state from source_name and destination_name (text inside brackets)
trip_df['source_state'] = trip_df['source_name'].str.extract(r'\((.*?)\)')
trip_df['destination_state'] = trip_df['destination_name'].str.extract(r'\((.*?)\)')

# Quick check
trip_df[['source_name', 'source_state', 'destination_name', 'destination_state']].head()
```

```python
print(trip_df['source_state'].isnull().sum())
print(trip_df['destination_state'].isnull().sum())
```

```python
print(sorted(trip_df['source_state'].dropna().unique()))
```

```python
trip_df['source_city_code'] = trip_df['source_name'].str.extract(r'^(.*?)\s*\(')[0].str.split(r'[_\s]').str[0]
trip_df['destination_city_code'] = trip_df['destination_name'].str.extract(r'^(.*?)\s*\(')[0].str.split(r'[_\s]').str[0]

trip_df[['source_name', 'source_city_code', 'destination_name', 'destination_city_code']].head()
```

```python
trip_df.to_csv('data/processed/trip_level_cleaned.csv', index=False)
print("Saved:", trip_df.shape)
```

```python
import pandas as pd

# Convert time columns to datetime first
trip_df['trip_creation_time'] = pd.to_datetime(trip_df['trip_creation_time'])
trip_df['od_start_time'] = pd.to_datetime(trip_df['od_start_time'])
trip_df['od_end_time'] = pd.to_datetime(trip_df['od_end_time'])

# 1. Trip Duration
trip_df['trip_duration_minutes'] = (trip_df['od_end_time'] - trip_df['od_start_time']).dt.total_seconds() / 60
trip_df['trip_duration_hours'] = trip_df['trip_duration_minutes'] / 60

# 2. OSRM Error (heart of the project)
trip_df['osrm_error'] = trip_df['actual_time'] - trip_df['osrm_time']

# 3. Delay Percentage
trip_df['delay_percentage'] = ((trip_df['actual_time'] - trip_df['osrm_time']) / trip_df['osrm_time']) * 100

# 4. Delay Flag (threshold: actual_time > osrm_time * 1.20)
trip_df['delay_flag'] = (trip_df['actual_time'] > trip_df['osrm_time'] * 1.20).astype(int)

# 5. Interstate Flag
trip_df['is_interstate'] = trip_df['source_state'] != trip_df['destination_state']

# 6. Corridor ID
trip_df['corridor_id'] = trip_df['source_state'] + '_' + trip_df['destination_state']

# 7. Trip Distance Category
bins = [0, 100, 300, 600, float('inf')]
labels = ['0-100km', '100-300km', '300-600km', '600+km']
trip_df['distance_category'] = pd.cut(trip_df['actual_distance_to_destination'], bins=bins, labels=labels)

# 8. Time Features
trip_df['creation_year'] = trip_df['trip_creation_time'].dt.year
trip_df['creation_month'] = trip_df['trip_creation_time'].dt.month
trip_df['creation_day'] = trip_df['trip_creation_time'].dt.day
trip_df['creation_day_name'] = trip_df['trip_creation_time'].dt.day_name()
trip_df['is_weekend'] = trip_df['trip_creation_time'].dt.dayofweek.isin([5, 6])
trip_df['creation_hour'] = trip_df['trip_creation_time'].dt.hour
trip_df['creation_quarter'] = trip_df['trip_creation_time'].dt.quarter

# 9. Route Efficiency (distance ratio)
trip_df['route_efficiency'] = trip_df['actual_distance_to_destination'] / trip_df['osrm_distance']

# 10. Time Efficiency
trip_df['time_efficiency'] = trip_df['actual_time'] / trip_df['osrm_time']

trip_df.shape
```

```python
validation_report = {
    'Original Rows (segments)': 144867,
    'Final Trips (after aggregation)': trip_df.shape[0],
    'Negative Distance': (trip_df['actual_distance_to_destination'] < 0).sum(),
    'Negative Time': (trip_df['actual_time'] < 0).sum(),
    'Duplicate trip_uuid': trip_df['trip_uuid'].duplicated().sum(),
    'Impossible Timestamps (end before start)': (trip_df['od_end_time'] < trip_df['od_start_time']).sum(),
    'Missing States': trip_df['source_state'].isnull().sum() + trip_df['destination_state'].isnull().sum(),
    'Missing Cities': trip_df['source_city_code'].isnull().sum() + trip_df['destination_city_code'].isnull().sum(),
    'Infinite/NaN in route_efficiency': trip_df['route_efficiency'].isin([float('inf'), float('-inf')]).sum() + trip_df['route_efficiency'].isnull().sum(),
    'Total Features Created': 10,
    'Final Columns': trip_df.shape[1]
}

validation_df = pd.DataFrame(list(validation_report.items()), columns=['Metric', 'Value'])
print(validation_df)
```

```python
# Full cleaned dataset (all columns, for reference/EDA)
trip_df.to_csv('data/processed/trip_level_cleaned.csv', index=False)

# ML-ready dataset (drop raw text/id columns not needed for modeling)
ml_ready_df = trip_df.drop(columns=['trip_uuid', 'route_schedule_uuid', 'source_center', 'destination_center',
                                      'source_name', 'destination_name', 'trip_creation_time',
                                      'od_start_time', 'od_end_time', 'cutoff_timestamp'], errors='ignore')
ml_ready_df.to_csv('data/processed/trip_level_ml_ready.csv', index=False)

validation_df.to_csv('data/processed/data_quality_report.csv', index=False)

print("Saved all 3 files. Final shape:", trip_df.shape)
```
